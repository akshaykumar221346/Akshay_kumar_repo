```python

# Class:
# Blueprint/template used to create objects.

# Object:
# Instance of a class.

# self:
# Refers to the current object.

# __init__:
# Automatically runs when an object is created and is commonly
# used to initialize object attributes.

# __str__:
# Human-readable representation of an object.

# __repr__:
# Developer-friendly representation useful for debugging.

# __dict__:
# Dictionary containing an object's attributes and values.

# Abstraction:
# Hides unnecessary implementation details and exposes
# essential functionality.

# Encapsulation:
# Keeps data and related methods together and can control
# access to internal data.

# Inheritance:
# Allows a child class to reuse properties and methods
# from a parent class.

# super():
# Gives access to methods of the parent class.

# Method Overriding:
# Child class provides its own version of a parent method.

# Positional Arguments:
# Arguments are assigned according to their position.

# Keyword Arguments:
# Arguments are assigned using parameter names.

# *args:
# Accepts multiple positional arguments.
# Stored as a tuple.

# **kwargs:
# Accepts multiple keyword arguments.
# Stored as a dictionary.
```

```python
# ============================================================
# PYTHON OOP (OBJECT-ORIENTED PROGRAMMING) NOTES
# ============================================================

# OOP stands for Object-Oriented Programming.
# OOP is a programming approach where we organize code using
# classes and objects.
#
# Main OOP concepts:
# 1. Class and Object
# 2. Encapsulation
# 3. Abstraction
# 4. Inheritance
# 5. Polymorphism


# ============================================================
# 1. CLASS AND OBJECT
# ============================================================

# A CLASS is a blueprint/template used to create objects.
# It defines the attributes (data) and methods (behavior).
#
# An OBJECT is an instance of a class.
#
# Example:
# Person = Class
# person1 = Object


class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print("Hello, my name is", self.name)


person1 = Person("Akshay", 25)
person2 = Person("John", 30)

print(person1.name)
print(person1.age)

person1.greet()


# ============================================================
# 2. __init__ METHOD
# ============================================================

# __init__ is a special method called automatically when
# an object is created.
#
# It is commonly used to initialize object attributes.
#
# self refers to the current object.


class Student:

    def __init__(self, name, course):
        self.name = name
        self.course = course


student1 = Student("Akshay", "Data Science")

print(student1.name)
print(student1.course)


```

```python
# ============================================================
# 3. SELF
# ============================================================

# self represents the current instance/object of the class.
#
# It is used to access variables and methods belonging
# to that particular object.
#
# self is written as the first parameter of instance methods.


class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print(self.name, self.salary)


emp1 = Employee("Akshay", 80000)

emp1.display()


# ============================================================
# 4. __str__ METHOD
# ============================================================

# __str__ is a special method that defines how an object
# should be displayed as a human-readable string.
#
# It is automatically called when we use:
# print(object)
# str(object)


class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person name is {self.name} and age is {self.age}"


person1 = Person("Akshay", 25)

print(person1)


```

```python
# ============================================================
# 5. __repr__ METHOD (REPRESENTATION)
# ============================================================

# __repr__ means representation.
#
# It gives a developer-friendly representation of an object.
#
# __repr__ is mainly useful for:
# debugging
# development
# logging
#
# Ideally, __repr__ should clearly show the object's
# important information.


class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Person(name='{self.name}', age={self.age})"


person1 = Person("Akshay", 25)

print(repr(person1))


# ============================================================
# 6. DIFFERENCE BETWEEN __str__ AND __repr__
# ============================================================

# __str__
# Used for users.
# Gives a readable representation.
# Called using print(object) or str(object).
#
# __repr__
# Used mainly for developers.
# Gives a detailed representation useful for debugging.
# Called using repr(object).


class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} costs ${self.price}"

    def __repr__(self):
        return f"Product(name='{self.name}', price={self.price})"


product1 = Product("Laptop", 1000)

print(str(product1))
print(repr(product1))


# ============================================================
# 7. __dict__
# ============================================================

# __dict__ is a dictionary containing an object's attributes
# and their current values.
#
# It is useful when we want to inspect the data stored
# inside an object.


class Person:

    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city


person1 = Person("Akshay", 25, "Baltimore")

print(person1.__dict__)

# Output will look like:
# {'name': 'Akshay', 'age': 25, 'city': 'Baltimore'}


# ============================================================
# 8. ABSTRACTION
# ============================================================

# Abstraction means hiding unnecessary implementation details
# and showing only the essential functionality to the user.
#
# Example:
# When we drive a car, we use the steering wheel, brake,
# and accelerator without knowing exactly how the engine works.
#
# In Python, abstraction can be implemented using the
# abc module and abstract methods.


from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car started")


car1 = Car()

car1.start()


# ============================================================
# 9. ENCAPSULATION
# ============================================================

# Encapsulation means keeping data and the methods that
# operate on that data together inside a class.
#
# It can also be used to control access to internal data.
#
# Python commonly uses:
#
# public    -> name
# protected -> _name
# private   -> __name
#
# Double underscore causes name mangling.
# It discourages direct access but is not absolute security.


class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount


account = BankAccount(1000)

account.deposit(500)

print(account.get_balance())

# Avoid:
# print(account.__balance)
#
# This produces an AttributeError because __balance
# is name-mangled.


# ============================================================
# 10. INHERITANCE
# ============================================================

# Inheritance allows one class to inherit attributes and
# methods from another class.
#
# Parent Class / Base Class:
# The class being inherited from.
#
# Child Class / Derived Class:
# The class that inherits from another class.
#
# Benefits:
# Code reuse
# Less duplication
# Easier maintenance


class Animal:

    def eat(self):
        print("Animal is eating")


class Dog(Animal):

    def bark(self):
        print("Dog is barking")


dog1 = Dog()

dog1.eat()
dog1.bark()


# ============================================================
# 11. INHERITANCE WITH __init__
# ============================================================

# super() is used to access methods from the parent class.
#
# super().__init__() is commonly used to initialize
# attributes defined in the parent class.


class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):

    def __init__(self, name, age, course):
        super().__init__(name, age)
        self.course = course

    def display(self):
        print(self.name, self.age, self.course)


student1 = Student("Akshay", 25, "Data Science")

student1.display()


# ============================================================
# 12. METHOD OVERRIDING
# ============================================================

# Method overriding happens when a child class defines
# a method with the same name as a method in the parent class.
#
# The child's version replaces/overrides the parent's version
# when called on the child object.
#
# Method overriding is an example of polymorphism.


class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Dog says Woof")


class Cat(Animal):

    def sound(self):
        print("Cat says Meow")


animal = Animal()
dog = Dog()
cat = Cat()

animal.sound()
dog.sound()
cat.sound()


```

```python
# ============================================================
# 13. POSITIONAL ARGUMENTS
# ============================================================

# Positional arguments are arguments passed based on
# their position/order.
#
# The order of values is important.


def student(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)


student("Akshay", 25, "Data Science")

# Here:
# "Akshay" -> name
# 25 -> age
# "Data Science" -> course


# ============================================================
# 14. KEYWORD ARGUMENTS
# ============================================================

# Keyword arguments are passed using parameter names.
#
# Because parameter names are specified, the order does
# not normally matter.


def student(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)


student(
    course="Data Science",
    name="Akshay",
    age=25
)


# ============================================================
# 15. POSITIONAL VS KEYWORD ARGUMENTS
# ============================================================

# Positional:
# Values are assigned based on their order.
#
# Keyword:
# Values are assigned based on parameter names.


def employee(name, salary):
    print(name, salary)


# Positional arguments
employee("Akshay", 80000)

# Keyword arguments
employee(salary=80000, name="Akshay")


# ============================================================
# 16. *args
# ============================================================

# *args allows a function to accept any number of
# positional arguments.
#
# args is stored as a TUPLE.
#
# The name "args" is conventional.
# The * is what provides the special behavior.


def addition(*args):

    print(args)

    total = 0

    for number in args:
        total += number

    return total


print(addition(10, 20))

print(addition(10, 20, 30, 40, 50))


# Another example:


def students(*names):

    for name in names:
        print(name)


students("Akshay", "John", "Sam", "David")


# ============================================================
# 17. **kwargs
# ============================================================

# **kwargs allows a function to accept any number of
# keyword arguments.
#
# kwargs is stored as a DICTIONARY.
#
# key   -> argument name
# value -> argument value


def person_details(**kwargs):

    print(kwargs)

    for key, value in kwargs.items():
        print(key, ":", value)


person_details(
    name="Akshay",
    age=25,
    course="Data Science"
)


# ============================================================
# 18. *args VS **kwargs
# ============================================================

# *args
# Accepts multiple positional arguments.
# Stored as a tuple.
#
# **kwargs
# Accepts multiple keyword arguments.
# Stored as a dictionary.


def details(*args, **kwargs):

    print("Positional arguments:", args)

    print("Keyword arguments:", kwargs)


details(
    "Akshay",
    25,
    course="Data Science",
    city="Baltimore"
)


# ============================================================
# 19. *args AND **kwargs IN A CLASS
# ============================================================


class Employee:

    def __init__(self, name, *skills, **details):

        self.name = name
        self.skills = skills
        self.details = details

    def display(self):

        print("Name:", self.name)
        print("Skills:", self.skills)
        print("Details:", self.details)


emp1 = Employee(
    "Akshay",
    "Python",
    "SQL",
    "Machine Learning",
    city="Baltimore",
    experience=2
)

emp1.display()


# ============================================================
# 20. BAD PRACTICES IN OOP
# ============================================================

# BAD PRACTICE 1:
# Creating classes with unclear names.


class A:
    pass


# Better:
class Customer:
    pass


# ------------------------------------------------------------

# BAD PRACTICE 2:
# Using unclear variable names.


class Person:

    def __init__(self, n, a):
        self.n = n
        self.a = a


# Better:


class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age


# ------------------------------------------------------------

# BAD PRACTICE 3:
# Creating attributes outside __init__ unnecessarily.


class Person:

    def set_name(self):
        self.name = "Akshay"


# Better:


class Person:

    def __init__(self, name):
        self.name = name


# ------------------------------------------------------------

# BAD PRACTICE 4:
# Repeating code instead of using inheritance.


class Dog:

    def eat(self):
        print("Eating")


class Cat:

    def eat(self):
        print("Eating")


# Better:


class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):
    pass


class Cat(Animal):
    pass


# ------------------------------------------------------------

# BAD PRACTICE 5:
# Accessing internal/private data directly when a class
# provides a proper method to interact with it.


class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance


account = BankAccount(1000)

print(account.get_balance())


# ============================================================
# 21. COMPLETE OOP EXAMPLE
# ============================================================

# This example combines:
# class
# object
# __init__
# __str__
# __repr__
# inheritance
# super()
# method overriding
# *args
# **kwargs


class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def work(self):
        print("Person is working")

    def __str__(self):
        return f"{self.name} is {self.age} years old"

    def __repr__(self):
        return f"Person(name='{self.name}', age={self.age})"


class Employee(Person):

    def __init__(self, name, age, *skills, **details):

        super().__init__(name, age)

        self.skills = skills
        self.details = details

    # Method overriding
    def work(self):
        print(self.name, "is working as an employee")

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Skills:", self.skills)
        print("Details:", self.details)


emp1 = Employee(
    "Akshay",
    25,
    "Python",
    "SQL",
    "Machine Learning",
    role="Data Scientist",
    city="Baltimore"
)

print(emp1)

print(repr(emp1))

emp1.work()

emp1.display()

print(emp1.__dict__)


```
